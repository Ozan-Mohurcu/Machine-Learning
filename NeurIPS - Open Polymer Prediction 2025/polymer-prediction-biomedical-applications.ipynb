```python
def discover_molecular_descriptors(max_autocorr_features=10):
    """
    Discovers and optimizes all available molecular descriptors from RDKit.
    
    Parameters:
    - max_autocorr features: Maximum number of AUTOCORR 2D descriptors
    
    Returns:
    - descriptor_functions: Descriptor functions
    - feature_names: Feature names
    """
    
    all_descriptors = []
    test_molecule = Chem.MolFromSmiles('CCO')  

    
    for descriptor_name in dir(Descriptors):
        if not descriptor_name.startswith('_'):
            try:
                descriptor_func = getattr(Descriptors, descriptor_name)
                if callable(descriptor_func):
                    result = descriptor_func(test_molecule)
                    # Numerik ve geçerli sonuçları filtrele
                    if isinstance(result, (int, float)) and not np.isnan(result):
                        all_descriptors.append((descriptor_name, descriptor_func))
            except:
                continue

    print(f"🔍 Total Discovered Identifiers: {len(all_descriptors)}")

    
    autocorr_descriptors = [(name, func) for name, func in all_descriptors 
                           if name.startswith('AUTOCORR2D_')]
    autocorr_descriptors.sort(key=lambda x: int(x[0].split('_')[-1]))
    
    
    selected_autocorr = autocorr_descriptors[:max_autocorr_features]
    
   
    other_descriptors = [(name, func) for name, func in all_descriptors 
                        if not name.startswith('AUTOCORR2D_')]
    
    
    final_descriptors = selected_autocorr + other_descriptors
    
    feature_names = [name for name, _ in final_descriptors]
    
    print(f"✅ Number of Optimized Descriptors: {len(final_descriptors)}")
    print(f"   (AUTOCORR2D: {len(selected_autocorr)}, Other: {len(other_descriptors)})")
    
    return final_descriptors, feature_names


molecular_descriptors = discover_molecular_descriptors(max_autocorr_features=10)
```

<h3 style="color: #e74c3c; text-shadow: 1px 1px 2px rgba(0,0,0,0.3); font-size: 20px;">🧮 Convert SMILES to Feature Matrix</h3>

```python
def extract_molecular_features(smiles_list, descriptor_functions):
    """
    Converts a SMILES list into a molecular feature matrix.
    
    Parameters:
    - smiles_list: List of SMILES notations
    - descriptor_functions: Descriptor functions
    
    Returns::
    - feature_matrix: Feature matrix in numpy array format
    """
    
    feature_matrix = []
    total_molecules = len(smiles_list)
    
    print(f"🔬 {total_molecules:,} molecule is being processed...", end="", flush=True)
    
    for idx, smiles in enumerate(smiles_list):
        
        if idx > 0 and (idx % 1000 == 0 or idx == total_molecules - 1):
            print(f" {idx+1:,}/{total_molecules:,}", end="", flush=True)
        
        molecule_features = []
        
        try:
            
            molecule = Chem.MolFromSmiles(smiles)
            
            if molecule is None:
                
                molecule_features = [np.nan] * len(descriptor_functions)
            else:
               
                for descriptor_name, descriptor_func in descriptor_functions:
                    try:
                        feature_value = descriptor_func(molecule)
                        
                        if np.isinf(feature_value) or abs(feature_value) > 1e10:
                            feature_value = np.nan
                        molecule_features.append(feature_value)
                    except:
                        
                        molecule_features.append(np.nan)
        except:
            
            molecule_features = [np.nan] * len(descriptor_functions)
        
        feature_matrix.append(molecule_features)
    
    print(" ✅ Completed", flush=True)
    return np.array(feature_matrix, dtype=float)


descriptor_functions, feature_names = molecular_descriptors
feature_matrix_raw = extract_molecular_features(combined_dataset['SMILES'].values, 
                                               descriptor_functions)
```

<div style="background-color: #b3e5fc; border: 2px solid #0288d1; border-radius: 12px; padding: 20px; margin-top: 20px; box-shadow: 3px 3px 8px rgba(0,0,0,0.2);">

<h3 style="color: #e74c3c; text-shadow: 1px 1px 2px rgba(0,0,0,0.3); font-size: 20px;">🧹 Data Cleaning and Preprocessing</h3>

```python
def clean_and_impute_features(feature_matrix):
    """
    Cleans missing values ​​and fills with median.
    
    Parameters:
    - feature_matrix: Raw feature matrix
    
    Returns:
    - cleaned_matrix: Cleaned feature matrix
    """
    
   
    cleaned_matrix = feature_matrix.copy()
    
    
    cleaned_matrix[np.isinf(cleaned_matrix)] = np.nan
    
    
    missing_count = np.isnan(cleaned_matrix).sum()
    missing_percentage = (missing_count / cleaned_matrix.size) * 100
    
    print(f"🧹 {missing_count:,} missing value cleared ({missing_percentage:.1f}%)")
    
    
    for column_idx in range(cleaned_matrix.shape[1]):
        column_data = cleaned_matrix[:, column_idx]
        if np.isnan(column_data).any():
            column_median = np.nanmedian(column_data)
            if not np.isnan(column_median):
                cleaned_matrix[np.isnan(column_data), column_idx] = column_median
            else:
                
                cleaned_matrix[np.isnan(column_data), column_idx] = 0
    
    return cleaned_matrix


cleaned_features = clean_and_impute_features(feature_matrix_raw)
```

<div style="background-color: #b3e5fc; border: 2px solid #0288d1; border-radius: 12px; padding: 20px; margin-top: 20px; box-shadow: 3px 3px 8px rgba(0,0,0,0.2);">

<h3 style="color: #e74c3c; text-shadow: 1px 1px 2px rgba(0,0,0,0.3); font-size: 20px;">🤖 Model Training - LightGBM Ensemble Approach</h3>

```python
def train_ensemble_model(X_features, y_target):
    """
    Trains an ensemble model for a given target variable.
    
    Parameters:
    - X_features: Feature matrix
    - y_target: Target variable
    
    Returns:
    - models: List of trained models
    - scaler: Feature scaler
    - cv_score: Cross-validation score
    - validation_data: Validation data
    """
    
    print(f"📊 Training Data: {len(y_target):,} sample")
    print(f"📈 Value Range: {y_target.min():.4f} - {y_target.max():.4f}")
    
   
    feature_scaler = StandardScaler()
    X_scaled = feature_scaler.fit_transform(X_features)
    
    
    lgb_params = {
        'objective': 'regression',
        'metric': 'mae',
        'boosting_type': 'gbdt',
        'num_leaves': 127,
        'learning_rate': 0.3,
        'feature_fraction': 0.8,
        'bagging_fraction': 0.9,
        'bagging_freq': 1,
        'lambda_l1': 0.1,
        'lambda_l2': 0.1,
        'min_data_in_leaf': 10,
        'verbose': -1,
        'random_state': 42
    }
    
    
    validation_scores = []
    trained_models = []
    all_val_true = []
    all_val_pred = []
    
    kfold = KFold(n_splits=2, shuffle=True, random_state=42)
    
    for fold_idx, (train_idx, val_idx) in enumerate(kfold.split(X_scaled)):
        
        X_train, X_val = X_scaled[train_idx], X_scaled[val_idx]
        y_train, y_val = y_target[train_idx], y_target[val_idx]
        
        
        train_data = lgb.Dataset(X_train, label=y_train)
        val_data = lgb.Dataset(X_val, label=y_val, reference=train_data)
        
        model = lgb.train(
            lgb_params,
            train_data,
            valid_sets=[val_data],
            num_boost_round=2000,
            callbacks=[lgb.early_stopping(50), lgb.log_evaluation(0)]
        )
        
        
        val_predictions = model.predict(X_val)
        fold_score = mean_absolute_error(y_val, val_predictions)
        
        validation_scores.append(fold_score)
        trained_models.append(model)
        
        
        all_val_true.extend(y_val)
        all_val_pred.extend(val_predictions)
        
        print(f"  Fold {fold_idx+1} completed - MAE: {fold_score:.4f}")
    
    
    mean_cv_score = np.mean(validation_scores)
    std_cv_score = np.std(validation_scores)
    
    print(f"🎯 CV Score: {mean_cv_score:.4f} ± {std_cv_score:.3f}")
    
    return trained_models, feature_scaler, mean_cv_score, all_val_true, all_val_pred
```

<div style="background-color: #b3e5fc; border: 2px solid #0288d1; border-radius: 12px; padding: 20px; margin-top: 20px; box-shadow: 3px 3px 8px rgba(0,0,0,0.2);">

<h3 style="color: #e74c3c; text-shadow: 1px 1px 2px rgba(0,0,0,0.3); font-size: 20px;">🔄 Training Cycle for All Target Variables</h3>

```python
trained_models_dict = {}
trained_scalers_dict = {}
cross_val_scores = []


all_validation_true = {}
all_validation_pred = {}

print("🚀 Model Training Process Begins...")
print("=" * 60)


for target_name in target_properties:
    print(f"\n🎯 {target_name} Model training for:")
    
    
    valid_mask = combined_dataset[target_name].notna()
    X_target = cleaned_features[valid_mask]
    y_target = combined_dataset[target_name].values[valid_mask]
    
   
    models, scaler, cv_score, val_true, val_pred = train_ensemble_model(X_target, y_target)
    
   
    trained_models_dict[target_name] = models
    trained_scalers_dict[target_name] = scaler
    cross_val_scores.append(cv_score)
    
    # Validasyon sonuçlarını kaydet
    all_validation_true[target_name] = val_true
    all_validation_pred[target_name] = val_pred
    
    print(f"✅ {target_name} model completed")

print("\n" + "=" * 60)
print("🎊 All Models Trained Successfully...")
```

<div style="background-color: #b3e5fc; border: 2px solid #0288d1; border-radius: 12px; padding: 20px; margin-top: 20px; box-shadow: 3px 3px 8px rgba(0,0,0,0.2);">

<h3 style="color: #e74c3c; text-shadow: 1px 1px 2px rgba(0,0,0,0.3); font-size: 20px;">📊 Model Performance Evaluation</h3>

```python
NULL_FOR_SUBMISSION = np.nan  

def weighted_mae(y_true, y_pred, ranges, counts, weights):
    maes = np.nanmean(np.abs(y_true - y_pred), axis=0)
    return np.sum(weights * maes)


ranges = []
counts = []
for target in target_properties:
    vals = np.array(all_validation_true[target])
    ranges.append(np.nanmax(vals) - np.nanmin(vals))
    counts.append(np.sum(~np.isnan(vals)))

ranges = np.array(ranges)
counts = np.array(counts)
counts[counts == 0] = 1
ranges[ranges == 0] = 1

weights = 1 / (np.sqrt(counts) * ranges)
weights /= weights.sum()

# Padding işlemi:
validation_true_df = pd.DataFrame()
validation_pred_df = pd.DataFrame()

max_length = max(len(all_validation_true[target]) for target in target_properties)

for target in target_properties:
    current_length = len(all_validation_true[target])
    padding_length = max_length - current_length

    true_padded = list(all_validation_true[target]) + [NULL_FOR_SUBMISSION] * padding_length
    pred_padded = list(all_validation_pred[target]) + [NULL_FOR_SUBMISSION] * padding_length

    validation_true_df[target] = true_padded
    validation_pred_df[target] = pred_padded

validation_true_df['id'] = range(len(validation_true_df))
validation_pred_df['id'] = range(len(validation_pred_df))


individual_scores = []
for i, target in enumerate(target_properties):
    y_true = np.array(all_validation_true[target])
    y_pred = np.array(all_validation_pred[target])
    score = np.nanmean(np.abs(y_true - y_pred)) * weights[i] * len(target_properties)
    individual_scores.append(score)


overall_competition_score = weighted_mae(
    validation_true_df[target_properties].values,
    validation_pred_df[target_properties].values,
    ranges,
    counts,
    weights
)

print("🏆 MODEL PERFORMANCE REPORT")
print("=" * 60)
print(f"Total Trained Model: {len(target_properties)} target × 2 fold = {len(target_properties) * 2}")
print(f"Average CV MAE: {np.mean(cross_val_scores):.4f}")
print(f"Individual Scores: {[f'{s:.4f}' for s in individual_scores]}")
print(f"🎯 ESTIMATED COMPETITION SCORE: {overall_competition_score:.4f}")
print("=" * 60)
```

<h3 style="color: #e74c3c; text-shadow: 1px 1px 2px rgba(0,0,0,0.3); font-size: 20px;">🔮 Prediction on Test Data</h3>

```python
def predict_target_properties(test_dataframe, target_name, models_list, scaler_obj):
    """
    Predicts target feature for test data.
    
    Parameters:
    - test_dataframe: Test dataset
    - target_name: Target variable name
    - models_list: Trained models
    - scaler_obj: Feature scaler
    
    Returns:
    - predictions: Prediction values
    """
    
    print(f"🔮 {target_name} It is estimated...")
    
    if models_list is None or scaler_obj is None:
        print(f"❌ {target_name} No model found for , returning zero values")
        return np.zeros(len(test_dataframe))
    
    # Moleküler özellikler
    descriptor_funcs, _ = molecular_descriptors
    test_features_raw = extract_molecular_features(test_dataframe['SMILES'].values, 
                                                  descriptor_funcs)
    test_features_clean = clean_and_impute_features(test_features_raw)
    
    # Ölçekleme
    test_features_scaled = scaler_obj.transform(test_features_clean)
    
    # Ensemble tahmini (tüm fold'lardan ortalama)
    fold_predictions = []
    for model in models_list:
        fold_pred = model.predict(test_features_scaled)
        fold_predictions.append(fold_pred)
    
    final_predictions = np.mean(fold_predictions, axis=0)
    
    print(f"📊 Tahmin aralığı: {final_predictions.min():.4f} - {final_predictions.max():.4f}")
    
    return final_predictions

# Test verisini yükle
test_dataset = pd.read_csv('/kaggle/input/neurips-open-polymer-prediction-2025/test.csv')

print("🧪 PREDICTION IS MADE ON TEST DATA")
print("=" * 50)


final_predictions = {}
for target in target_properties:
    predictions = predict_target_properties(
        test_dataset, 
        target, 
        trained_models_dict[target], 
        trained_scalers_dict[target]
    )
    final_predictions[target] = predictions
```
